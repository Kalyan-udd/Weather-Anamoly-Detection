```python
import pandas as pd
```

```python
df = pd.read_csv("data/Testing_data.csv")
df
```

```python
df["6_hour_gradient_temp"] = df['temperature'].diff(periods=6)
df["6_hour_gradient_press"] = df['pressure'].diff(periods=6)
df["6_hour_gradient_humid"] = df['humidity'].diff(periods=6)
df = df.dropna(axis=0).reset_index()
```

```python
df.drop(columns=['hour', 'month'], inplace=True)
```
