```python
data = df[['Date Time', 'p (mbar)', 'T (degC)','rh (%)']]
data.rename(columns={
       'p (mbar)': "pressure", 
       'T (degC)':"temperature",
       'rh (%)': "relative_humidity"
}, inplace=True)
```

```python
import matplotlib.pyplot as plot
```

```python
sample = data
plot.figure(figsize=(9, 5))
plot.scatter(
    sample['temperature'], 
    sample['relative_humidity'], 
    color='royalblue', 
    alpha=0.6, 
    edgecolors='none', 
    s=25
)

plot.title("Sensor Correlation: Temperature vs. Relative Humidity")
plot.xlabel("Temperature (°C)")
plot.ylabel("Relative Humidity (%)")
plot.grid(True, linestyle='--', alpha=0.5)
plot.show()
```
