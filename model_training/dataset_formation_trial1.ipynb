```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/gradient_data_kolkata.csv", parse_dates=['date']).set_index('date')
df
```

```python
df['hour'] = df.index.hour
df['month'] = df.index.month
df
```

converting raw hour and month numbers into some continuous cycle to have it continuous because as the 0hrs is exactly 1 hour from 23 hrs we could tell that by adding the cos and sine functions

```python
df['cos_hour'] = np.cos(2*(np.pi)*df['hour']/24)
df['sin_hour'] = np.sin(2*np.pi*df['hour']/24)
df['cos_month'] = np.cos(2*np.pi*df['month']/12)
df['sin_month'] = np.sin(2*np.pi*df['month']/12)
df
```

Before standardisation as we can notice the temperature and humidity and pressure columns have medium, large, and low gradient values which will eventually make the model to consider the effect of humidity more on anomaly and less of pressure so we need to convert the columns into some standard scaled values since this is a timeseries and if we use StandardScaler from sklearn we would get one mean for the overall 8 years of data not for every month for every hour. so we need to standardise the data with considering the hour mean and standard deviation for every month.

```python
df.to_csv("data/gradient_data_kolkata.csv")
```
